# Dataset Overview

*The dataset contains 371,528 rows and 20 columns.*

**Missing Values:**

Some columns contain missing values, including vehicleType, gearbox, model, fuelType, and notRepairedDamage.

**Data Types:**

The dataset contains both numerical and categorical data, and some categorical values are written in German.

```python
autos.columns
```

```python
def clean_column_name(column):
    new_name = ""

    for char in column:
        if char.isupper():
            new_name += "_" + char.lower()
        else:
            new_name += char.lower()

    return new_name


# print(clean_column_name("dateCrawled"))
# print(clean_column_name("yearOfRegistration"))
# print(clean_column_name("notRepairedDamage"))

autos.columns = autos.columns.map(clean_column_name)
autos.rename(columns={"power_p_s": "power_ps"}, inplace=True)
```
